```python
from cimgraph.databases import XMLFile
from cimgraph.models import FeederModel
import cimgraph.utils as utils
```

```python
import os
import cimgraph.data_profile.cimhub_2023 as cim
os.environ['CIMG_CIM_PROFILE'] = 'cimhub_2023'
# os.environ['CIMG_CIM_PROFILE'] = 'cimvfm.data_profiles.connectivity.switch'
# import cimvfm.data_profiles.connectivity.switch as cim
# os.environ['CIMG_VALIDATION_LOG_LEVEL'] = 'DEBUG'
```

```python
file = XMLFile(filename='/home/ande188/EASY-CIM/tests/IEEE13_Assets.xml')
network = FeederModel(container=cim.Feeder(), connection=file)
```

```python
from pathlib import Path
from cimgraph.models import GraphModel
from easycim.shacl.asset_catalog_builder import AssetCatalogBuilder

# Assuming you have a GraphModel with equipment data loaded
# network = GraphModel(database, feeder)  # Your existing network

# Create the catalog builder
catalog_builder = AssetCatalogBuilder(
    network=network,
    catalog_output_dir=Path('./my_equipment_catalogs'),
    shacl_shapes_file='./shapes/catalog_export_shapes.ttl'  # Optional - uses default if not provided
)

# Create catalogs for all equipment types
catalog_files = catalog_builder.create_equipment_catalogs()

print("Created catalog files:")
for equipment_type, file_path in catalog_files.items():
    print(f"  {equipment_type}: {file_path}")
```
