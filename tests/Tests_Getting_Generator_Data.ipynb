```python
import importlib
import logging
from cimgraph.models import GraphModel
from easycim.data_iterator import get_data
from easycim.reduced_data_profile import ReducedDataProfile
from cimgraph.models import FeederModel
from cimgraph.databases import ConnectionParameters, XMLFile
import cimgraph.data_profile.cimhub_2023 as cim

params = ConnectionParameters(filename="IEEE9500bal.xml", cim_profile='cimhub_2023', iec61970_301=8)
# params = ConnectionParameters(filename="IEEE13_Assets.xml", cim_profile='cimhub_2023', iec61970_301=8)
rdf = XMLFile(params)
#file = XMLFile(params) # file read connection
network = FeederModel(container=cim.Feeder(),connection=rdf) # create feeder model"
```

```python
data_profile = ReducedDataProfile()
cim_profile = network.connection.connection_params.cim_profile
cim = importlib.import_module(f'cimgraph.data_profile.{cim_profile}')

count = 0
generator_data = {}
if cim.SynchronousMachine in network.graph:
    for switch_equip in network.graph[cim.SynchronousMachine].values():
        generator_data[switch_equip.mRID] = get_data(switch_equip, data_profile.SynchronousMachine)
        #  generator_data[switch_equip.mRID] = get_data(switch_equip, ['ikk', 'maxQ', 'minQ', 'operatingMode', 'type'])
        count = count  + 1

print(count)
print(generator_data.keys())
print(generator_data.values())
        #     switch_data[switch_equip.mRID] = get_data(switch_equip, data_profile.LoadBreakSwitch)
```
